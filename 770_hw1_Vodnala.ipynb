## Task 2: standardization

```python
from sklearn.preprocessing import StandardScaler

# Creating StandardScaler
scaler = StandardScaler()

# Select only feature columns (excluding target column)
features = df_data.columns[:-1]

# Fit the scaler to features and transform them
df_data2[features] = scaler.fit_transform(df_data2[features])

# Renaming to data_b 
data_b = df_data2

# Display the first few rows
data_b.head()
```

# Task 3: standardization with Median 

```python
from sklearn.preprocessing import RobustScaler

# Create a RobustScaler
robust_scaler = RobustScaler()

# Select only feature columns (excluding target column)
features = df_data.columns[:-1]

# Fit RobustScaler to features and transform them
data_c = df_data.copy()  # copy to maintain the original data
data_c[features] = robust_scaler.fit_transform(data_c[features])

# Display the first few rows
data_c.head()
```

# Task 4: Split dataset into train_x, train_y, test_x and test_y

```python
df_data.shape
```

```python
from sklearn.model_selection import train_test_split

# Define the features and the target variable
X = data_c.drop('class', axis=1)  # Features
y = data_c['class']               # Target variable

# Split the data into training and test sets (65% training, 35% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.35, random_state=123)

# Display the shapes of the train and test sets to verify the sizes
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

# Task 5: Feature Selection (RFECV)

```python
from sklearn.svm import SVC
from sklearn.feature_selection import RFECV
from sklearn.metrics import make_scorer, roc_auc_score

# SVC estimator with linear kernel
svc = SVC(kernel='linear')

# Create the RFECV object with the SVC estimator and using roc_auc as the scoring method
rfecv = RFECV(estimator=svc, scoring='roc_auc')

# Fit RFECV to the training data using fit method (not fit_transform)
rfecv.fit(X_train, y_train)

# Printing features that are selected with their ranking
print("Features selected (True = selected):", rfecv.support_)
print("Feature ranking (1 = selected, >1 = not selected):", rfecv.ranking_)

# Transform the training and test sets to keep only the selected features
X_train_selected = X_train.loc[:, rfecv.support_]
X_test_selected = X_test.loc[:, rfecv.support_]

# Display the shapes of the new datasets to verify reduction in feature count
print("Shape of X_train_selected:", X_train_selected.shape)
print("Shape of X_test_selected:", X_test_selected.shape)
```

# Task 6: Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score

# Logistic Regression classifier
log_reg = LogisticRegression()

# Fit the model to training data
log_reg.fit(X_train_selected, y_train)

# Predicting with test set
y_pred = log_reg.predict(X_test_selected)

# Printing classification report
print("Classification Report:\n", classification_report(y_test, y_pred))

# accuracy of the model
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

# Task 7: SVM Model Training and Hyperparameter Tuning


#7.1 Explanation of SVM Parameters C and Kernel

In the SVM model, the "C" parameter is a crucial setting that helps manage the balance between creating a model that accurately captures the training data and one that remains efficient and generalized for new, unseen data. When the C value is high, the model strives to fit the training set as perfectly as possible, which means it's less forgiving of errors during the learning process. on the other hand, A lower C value allows the model more flexibility by letting it make more mistakes. This can be beneficial because it enables the model to adapt better when it encounters new or different data than what it was trained on, potentially improving its overall performance.

The "kernel" in an SVM plays a role in how the model views or interprets the data. Essentially, a kernel function takes the original data and transforms it into a new format that might be easier for the model to understand and separate into classes, especially when the groups are not easily distinguishable in the original format. For instance, a linear kernel doesn’t change the data it works well when the data is already easily separable. A polynomial kernel, however, transforms the data into a polynomial format which helps to distinguish datasets that have a more complex relationship. The Radial Basis Function (RBF) kernel, popular for handling non-linear data, considers the distance between data points, essentially grouping data that are close together. Lastly, the sigmoid kernel, similar to functions used in neural networks, transforms the data in a way that mimics the firing patterns of neurons in the brain.

```python
#7.2 Hyperparameter Tuning with GridSearchCV

from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC

# Setting up the parameter grid to tune
param_grid = {
    'C': [0.1, 1, 10, 50, 100],
    'kernel': ['poly', 'rbf', 'sigmoid']
}

#SVC classifier
svc = SVC()

# Create the GridSearchCV
grid_search = GridSearchCV(estimator=svc, param_grid=param_grid)

# Fit GridSearchCV
grid_search.fit(X_train_selected, y_train)

# Print the best estimator
print("Best estimator:", grid_search.best_estimator_)
```

```python
#7.3 Prediction and Evaluation Using the Best Model

# Retrieve the best model from grid search
best_svc = grid_search.best_estimator_

# Prediction on test data using best model
y_pred_svc = best_svc.predict(X_test_selected)

# Printing the classification report
from sklearn.metrics import classification_report, accuracy_score
print("Classification Report for SVM:\n", classification_report(y_test, y_pred_svc))

# Print the accuracy
print("Accuracy of SVM:", accuracy_score(y_test, y_pred_svc))
```
